```python
import pandas as pd
```

```python
# =====================================================
# Load Dataset & Basic Preprocessing
# =====================================================

df = pd.read_csv("../Raw Data/customers.csv")
df['signup_date'] = pd.to_datetime(df['signup_date'])
df.head()
```

```python
df.sample(10)
```

```python
print("Total Rows In The Dataset:",df.shape[0])
print("Total Columns In The Dataset:",df.shape[1])
```

```python
df.info()
```

```python
print("Total Missing Values In Columns:")
missing = df.isnull().sum()
missing_pct = (df.isnull().sum() / len(df) * 100).round(2)
    
result = pd.DataFrame({
    "missing_count": missing,
    "missing_percent": missing_pct
})
    
print(result[result["missing_count"] > 0])
```

```python
print("Total Duplicates Rows In customer_id:",df.duplicated(subset=["customer_id"]).sum())
```

```python
df['email'] = df['email'].fillna('Unknown')
df['phone'] = df['phone'].fillna(-1)

df['city'] = df['city'].fillna('Unknown')
df['state'] = df['state'].fillna('Unknown')
df['pincode'] = df['pincode'].fillna(-1)
```

```python
df['phone'] = df['phone'].astype('Int64')
df['pincode'] = df['pincode'].astype('Int64')
```

```python
# String Cleaning:
df['gender'] = (df["gender"]
                .str.strip()
                .str.lower()
                .str.title()
)

df['state'] = (df["state"]
                .str.strip()
                .str.lower()
                .str.title()
)

df['city'] = (df["city"]
                .str.strip()
                .str.lower()
                .str.title()
)

city_map = {"Banglore": "Bengaluru", "Gurgaon": "Gurugram", "Baroda": "Vadodara",
              "Calcutta": "Kolkata", "Allahabad": "Prayagraj"}
df['city'] = df['city'].replace(city_map)
```

```python
df["gender"].value_counts()
```

```python
print("Total State:", df["state"].nunique(),"\n")
print(df["state"].value_counts())
```

```python
print("Total City:", df["city"].nunique(),"\n")
print(df["city"].value_counts())
```

```python
# Detect near-duplicate customers (similar name + phone) with different customer_id
dupe_phone = df[df.duplicated(subset=['phone'], keep=False) & (df['phone'] != -1)]
print("Customers sharing phone numbers:", dupe_phone.shape[0])
dupe_phone
```
