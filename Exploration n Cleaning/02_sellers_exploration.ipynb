```python
import pandas as pd
```

```python
# =====================================================
# Load Dataset & Basic Preprocessing
# =====================================================

df = pd.read_csv("../Raw Data/sellers.csv")
df['seller_signup_date'] = pd.to_datetime(df['seller_signup_date'])
df.head()
```

```python
df.sample(10)
```

```python
print("Total Rows In The Dataset:",df.shape[0])
print("Total Columns In The Dataset:",df.shape[1])
```

```python
df.info()
```

```python
df.describe()
```

```python
print("Total Missing Values In Columns:")
missing = df.isnull().sum()
missing_pct = (df.isnull().sum() / len(df) * 100).round(2)
    
result = pd.DataFrame({
    "missing_count": missing,
    "missing_percent": missing_pct
})
    
print(result[result["missing_count"] > 0])
```

```python
print("Total Duplicates Rows In seller_id:",df.duplicated(subset=["seller_id"]).sum())
```

```python
df['seller_rating'] = df['seller_rating'].fillna(
    df.groupby('seller_tier')['seller_rating'].transform('median')
)
```

```python
# String Cleaning:
df['seller_name'] = (df["seller_name"]
                .str.strip()
                .str.lower()
                .str.title()
)

df['seller_state'] = (df["seller_state"]
                .str.strip()
                .str.lower()
                .str.title()
)

df['seller_city'] = (df["seller_city"]
                .str.strip()
                .str.lower()
                .str.title()
)

df['primary_category'] = (df["primary_category"]
                .str.strip()
                .str.lower()
                .str.title()
)
```

```python
print("Total State:", df["seller_state"].nunique(),"\n")
print(df["seller_state"].value_counts())
```

```python
print("Total City:", df["seller_city"].nunique(),"\n")
print(df["seller_city"].value_counts())
```

```python
print("Total Category:", df["primary_category"].nunique(),"\n")
print(df["primary_category"].value_counts())
```
