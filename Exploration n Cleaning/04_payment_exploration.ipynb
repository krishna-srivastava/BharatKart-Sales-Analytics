```python
import pandas as pd
```

```python
# =====================================================
# Load Dataset & Basic Preprocessing
# =====================================================

df = pd.read_csv("../Raw Data/payment.csv")
df['payment_date'] = pd.to_datetime(df['payment_date'])
df.head()
```

```python
df.sample(10)
```

```python
print("Total Rows In The Dataset:",df.shape[0])
print("Total Columns In The Dataset:",df.shape[1])
```

```python
df.info()
```

```python
df.describe()
```

```python
# String Cleaning:
df['payment_method'] = (df["payment_method"]
                .str.strip()
                .str.lower()
                .str.title()
)

df['payment_status'] = (df["payment_status"]
                .str.strip()
                .str.lower()
                .str.title()
)
df['payment_method'] = df['payment_method'].replace({'Upi': 'UPI', 'Cod': 'COD'})
```

```python
print("Total Missing Values In Columns:")
missing = df.isnull().sum()
missing_pct = (df.isnull().sum() / len(df) * 100).round(2)
    
result = pd.DataFrame({
    "missing_count": missing,
    "missing_percent": missing_pct
})
    
print(result[result["missing_count"] > 0])
```

```python
print("Total Duplicates Rows In payment_id:",df.duplicated(subset=["payment_id"]).sum())
print("Total Duplicates Rows In order_id:",df.duplicated(subset=["order_id"]).sum())
```

```python
print("Total payment_method:", df["payment_method"].nunique(),"\n")
print(df["payment_method"].value_counts())
```

```python
print("Payment Status:", df["payment_status"].nunique(),"\n")
print(df["payment_status"].value_counts())
```

```python
df.to_csv("../Clean Data/payment.csv",index=False)
```
