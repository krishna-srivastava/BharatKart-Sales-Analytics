```python
import pandas as pd
```

```python
# =====================================================
# Load Dataset & Basic Preprocessing
# =====================================================

df = pd.read_csv("../Raw Data/return.csv")
df['return_date'] = pd.to_datetime(df['return_date'])
df.head()
```

```python
df.sample(10)
```

```python
print("Total Rows In The Dataset:",df.shape[0])
print("Total Columns In The Dataset:",df.shape[1])
```

```python
df.rename(columns={'refund_amount': 'return_claim_amount'}, inplace=True)
df['actual_refund'] = df.apply(
    lambda x: x['return_claim_amount'] if x['return_status'] == 'Refunded' else 0, 
    axis=1
)
```

```python
df.info()
```

```python
df.describe()
```

```python
print("Total Missing Values In Columns:")
missing = df.isnull().sum()
missing_pct = (df.isnull().sum() / len(df) * 100).round(2)
    
result = pd.DataFrame({
    "missing_count": missing,
    "missing_percent": missing_pct
})
    
print(result[result["missing_count"] > 0])
```

```python
print("Total Duplicates Rows In return_id:",df.duplicated(subset=["return_id"]).sum())
print("Total Duplicates Rows In order_id:",df.duplicated(subset=["order_id"]).sum())
```

```python
print("Return Status:", df["return_status"].nunique(),"\n")
print(df["return_status"].value_counts())
```

```python
df.to_csv("../Clean Data/returns.csv",index=False)
```
