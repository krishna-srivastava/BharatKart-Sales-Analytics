```python
import pandas as pd
```

```python
# =====================================================
# Load Dataset & Basic Preprocessing
# =====================================================

df = pd.read_csv("../Raw Data/orders.csv")
df['order_date'] = pd.to_datetime(df['order_date'])
df['shipped_date'] = pd.to_datetime(df['shipped_date'])
df['delivery_date'] = pd.to_datetime(df['delivery_date'])
df['expected_delivery_date'] = pd.to_datetime(df['expected_delivery_date'])

df.head()
```

```python
df.info()
```

```python
print("Total Missing Values In Columns:")
missing = df.isnull().sum()
missing_pct = (df.isnull().sum() / len(df) * 100).round(2)
    
result = pd.DataFrame({
    "missing_count": missing,
    "missing_percent": missing_pct
})
    
print(result[result["missing_count"] > 0])
```

```python
df.describe()
```

```python
# String Cleaning:
df['order_status'] = (df["order_status"]
                .str.strip()
                .str.lower()
                .str.title()
)

df['cancellation_reason'] = (df["cancellation_reason"]
                .str.strip()
                .str.lower()
                .str.title()
)

df['delivery_city'] = (df["delivery_city"]
                .str.strip()
                .str.lower()
                .str.title()
)

df['delivery_state'] = (df["delivery_state"]
                .str.strip()
                .str.lower()
                .str.title()
)
```

```python
print("Total Duplicates Rows In order_id:",df.duplicated(subset=["order_id"]).sum())
```

```python
print("Total State:", df["delivery_state"].nunique(),"\n")
print(df["delivery_state"].value_counts())
```

```python
print("Total City:", df["delivery_city"].nunique(),"\n")
print(df["delivery_city"].value_counts())
```

```python
print("Order Status:", df["order_status"].nunique(),"\n")
print(df["order_status"].value_counts())  
```

```python
# Delivery performance metrics
df['delivery_days'] = (df['delivery_date'] - df['shipped_date']).dt.days
df['delivery_delay'] = (df['delivery_date'] - df['expected_delivery_date']).dt.days
```

```python
customer_orders = df.groupby('customer_id').agg(
    total_orders=('order_id', 'count'),
    total_spend=('order_total', 'sum'),
    last_order_date=('order_date', 'max')
).reset_index()

def segment(row):
    if row['total_orders'] == 1:
        return 'New'
    elif row['total_spend'] >= customer_orders['total_spend'].quantile(0.99):
        return 'VIP'
    else:
        return 'Repeat'

customer_orders['customer_segment'] = customer_orders.apply(segment, axis=1)

# merge back to orders df
df = df.merge(customer_orders[['customer_id', 'customer_segment']], on='customer_id', how='left')
customer_orders['customer_segment'].value_counts()
```
