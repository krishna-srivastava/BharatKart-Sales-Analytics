```python
import pandas as pd
```

```python
# =====================================================
# Load Dataset & Basic Preprocessing
# =====================================================

df = pd.read_csv("../Raw Data/order_items.csv")
df.head(10)
```

```python
df.sample(10)
```

```python
print("Total Rows In The Dataset:",df.shape[0])
print("Total Columns In The Dataset:",df.shape[1])
```

```python
df.info()
```

```python
df.describe()
```

```python
print("Total Missing Values In Columns:")
missing = df.isnull().sum()
missing_pct = (df.isnull().sum() / len(df) * 100).round(2)
    
result = pd.DataFrame({
    "missing_count": missing,
    "missing_percent": missing_pct
})
    
print(result[result["missing_count"] > 0])
```

```python
print("Total Duplicates Rows In order_id:",df.duplicated(subset=["order_item_id"]).sum())
```

```python
products = pd.read_csv("../Raw Data/products.csv")[['product_id', 'cost_price']]
df = df.merge(products, on='product_id', how='left')

df['profit'] = (df['item_total'] - (df['cost_price'] * df['quantity'])).round(2)

print("Negative profit rows (loss-making sales):", (df['profit'] < 0).sum())
df.sample(10)
```

```python
loss_rows = df[df['profit'] < 0]
print("Avg discount on loss-making sales:", loss_rows['discount_percent'].mean())
print("Avg discount overall:", df['discount_percent'].mean())
```

```python
df.to_csv("../Clean Data/order_items.csv",index=False)
```
