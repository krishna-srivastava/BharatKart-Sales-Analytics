```python
import pandas as pd

customers = pd.read_csv("../Clean Data/customers.csv")
sellers = pd.read_csv("../Clean Data/sellers.csv")
products = pd.read_csv("../Clean Data/products.csv")
orders = pd.read_csv("../Clean Data/orders.csv")
order_items = pd.read_csv("../Clean Data/order_items.csv")
payments = pd.read_csv("../Clean Data/payment.csv")
returns = pd.read_csv("../Clean Data/return.csv")
```

```python
pk_checks = {
    "customers": ("customer_id", customers),
    "sellers": ("seller_id", sellers),
    "products": ("product_id", products),
    "orders": ("order_id", orders),
    "order_items": ("order_item_id", order_items),
    "payments": ("payment_id", payments),
    "returns": ("return_id", returns)
}

for table, (pk, df) in pk_checks.items():
    print(
        table,
        "| rows:", len(df),
        "| unique PK:", df[pk].nunique(),
        "| duplicates:", df[pk].duplicated().sum()
    )
```

```python
# Customers → Orders:
invalid = orders[
    ~orders["customer_id"].isin(customers["customer_id"])
]

print("Invalid customer_id:", len(invalid))
```

```python
# Orders → Order Items:
invalid = order_items[
    ~order_items["order_id"].isin(orders["order_id"])
]

print("Invalid order_id:", len(invalid))
```

```python
# Products → Order Items:
invalid = order_items[
    ~order_items["product_id"].isin(products["product_id"])
]

print("Invalid product_id:", len(invalid))
```

```python
# Sellers → Order Items:
invalid = order_items[
    ~order_items["seller_id"].isin(sellers["seller_id"])
]

print("Invalid seller_id:", len(invalid))
```

```python
# Orders → Payments:
invalid = payments[
    ~payments["order_id"].isin(orders["order_id"])
]

print("Invalid payment order_id:", len(invalid))

print("Duplicate order_id in payments:",
      payments["order_id"].duplicated().sum())
```

```python
# Order Items → Returns:
invalid = returns[
    returns["order_item_id"].notna() &
    ~returns["order_item_id"].isin(order_items["order_item_id"])
]

print("Invalid order_item_id:", len(invalid))
```

```python
# Orders → Returns:
invalid = returns[
    ~returns["order_id"].isin(orders["order_id"])
]

print("Invalid return order_id:", len(invalid))
```
